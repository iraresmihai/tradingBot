```python
class InfoCreate():
    def __init__(self, file_paths, symbols, start, end, whereToSave, spread):
        self.file_paths = file_paths
        self.symbols = symbols
        self.start = start
        self.end = end

    def get_data(self):
        frames = []
        for name in file_paths:
            frames.append(pd.read_parquet(name))
        aux = pd.concat(frames).sort_index()
        aux = aux[aux.index.get_level_values("symbol").isin(symbols)]
        aux = aux[~aux.index.duplicated(keep="last")]
        aux = aux.unstack("symbol").swaplevel(axis=1).sort_index(axis=1)

        self.data = aux.loc[self.start:self.end].copy().ffill()
        names = self.data.columns.get_level_values("symbols").unique()
        for name in names:
            self.data[name]["bid"] = self.data[name]["open"] + spread
            self.data[name]["ask"] = self.data[name]["open"] - spread
```
